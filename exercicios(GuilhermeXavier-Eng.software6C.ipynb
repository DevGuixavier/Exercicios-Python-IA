```python
# Declaramos n como a variável que recebe o número.
# Assumimos e_primo como True.
# Depois usamos um range começando em 2 até n,
# e vamos fazendo a divisão de n pelos números percorridos nesse intervalo.
# Se ele travar em qualquer uma dessas divisões que deixe resto zero,
# ou seja, se o que sobra quando divide n por i for igual a 0,
# significa que não é primo.
# Quando isso acontece, muda e_primo para False e usa break para sair do laço.
# Se o laço termina sem encontrar nenhuma divisão que deixe resto zero,
# e_primo continua True, confirmando que o número é primo.


n = int(input('digite um número: '))
e_primo = True

for i in range(2, n):
    if n % i == 0:
        e_primo = False
        break

if e_primo:
    print(f'{n} é primo')
else:
    print(f'{n} não é primo')
```

## 20. Matriz Transposta
Dada uma matriz (lista de listas), crie sua matriz transposta.

```python
# Defino uma matriz de exemplo (uma lista que contém outras listas dentro).
# Crio uma lista vazia chamada 'transposta' para guardar o resultado.
# Crio um laço para percorrer as colunas da matriz original (baseado no tamanho da primeira linha).
# Para cada coluna, crio uma 'nova_linha' vazia.
# Uso um segundo laço para percorrer as linhas originais e pegar o item correspondente àquela coluna.
# Adiciono a 'nova_linha' na matriz transposta e, no final, imprimo o resultado.

matriz = [
    [1, 2, 3],
    [4, 5, 6]
]

transposta = []

for i in range(len(matriz[0])):
    nova_linha = []
    for linha in matriz:
        nova_linha.append(linha[i])
    transposta.append(nova_linha)

print("Matriz Original:")
for linha in matriz:
    print(linha)

print("Matriz Transposta:")
for linha in transposta:
    print(linha)
```

## 21. Busca Binária
Implemente a busca binária para encontrar um elemento em uma lista ordenada.

```python
# Defino uma lista ordenada e o número alvo que quero encontrar.
# Crio 'inicio' no índice 0 e 'fim' como 'len(lista) - 1' para pegar o índice do último elemento.
# Crio um laço 'while' que roda enquanto a faixa de busca for válida (inicio <= fim).
# Calculo o índice do 'meio' e comparo o valor dessa posição com o alvo.
# Se acertar, exibo o resultado e saio do laço com 'break'.
# Se o número do meio for menor/maior que o alvo, ajusto 'inicio' ou 'fim' para descartar a metade errada da lista.
# Se o laço terminar sem encontrar o número, aviso que ele não está na lista.

lista = [10, 20, 30, 40, 50, 60, 70, 80, 90]
alvo = 70

inicio = 0
fim = len(lista) - 1
encontrado = False

while inicio <= fim:
    meio = (inicio + fim) // 2

    if lista[meio] == alvo:
        print(f'Elemento {alvo} encontrado no índice {meio}!')
        encontrado = True
        break
    elif lista[meio] < alvo:
        inicio = meio + 1
    else:
        fim = meio - 1

if not encontrado:
    print(f'Elemento {alvo} não foi encontrado na lista.')
```

## 22. Criptografia de Cifra de César
Crie uma função para criptografar uma mensagem usando a cifra de César.

```python
# Crio a função que recebe a mensagem e a 'chave' (o número de casas a avançar no alfabeto).
# Defino a string com o alfabeto e crio 'resultado' vazia para ir guardando as novas letras.
# Percorro caractere por caractere: se for uma letra do alfabeto, acho a posição dela usando .find().
# Somo a 'chave' à posição para andar as casas para a frente e uso o resto da divisão (% 26) para o Z voltar para o A.
# Mantenho o formato maiúsculo se a letra original for maiúscula e não altero espaços ou pontuações.
# Defino uma variável casas = 3'-> ou seja, cada letra vai andar 3 casas para a frente, fazendo o 'A' virar 'D' e assim por diante.

def criptografar_cesar(mensagem, deslocamento):
    alfabeto = "abcdefghijklmnopqrstuvwxyz"
    resultado = ""

    for letra in mensagem:
        letra_minuscula = letra.lower()

        if letra_minuscula in alfabeto:

            posicao = alfabeto.find(letra_minuscula)


            nova_posicao = (posicao + deslocamento) % 26
            nova_letra = alfabeto[nova_posicao]


            if letra.isupper():
                resultado += nova_letra.upper()
            else:
                resultado += nova_letra
        else:
            resultado += letra

    return resultado


texto = "senhadomeucelulare12345!"
casas = 3
texto_cifrado = criptografar_cesar(texto, casas)

print(f"Original: {texto}")
print(f"Cifrado:  {texto_cifrado}")
```

## 23. Descriptografar Cifra de César
Crie uma função para descriptografar uma mensagem cifrada com a cifra de César.

```python
# Aproveitando a frase que coloquei para criptografar no exercício acima, reutilizei ela nesse aqui
# Crio a função que recebe a mensagem cifrada e o 'deslocamento' (o número de casas a voltar no alfabeto).
# Defino a string com o alfabeto e crio 'resultado' vazia para ir guardando o texto original.
# Percorro caractere por caractere: se for uma letra do alfabeto, acho a posição dela usando .find().
# Subtraio o 'deslocamento' da posição para voltar as casas e uso o resto da divisão (% 26) para o A voltar para o Z.
# Mantenho o formato maiúsculo se a letra for maiúscula e não altero espaços ou pontuações.
# utilizo a várivel 'casas = 3' -> a mesma quantidade de casas usada para criptografar, para conseguir decifrar a mensagem.

def descriptografar_cesar(mensagem, deslocamento):
    alfabeto = "abcdefghijklmnopqrstuvwxyz"
    resultado = ""

    for letra in mensagem:
        letra_minuscula = letra.lower()

        if letra_minuscula in alfabeto:
            posicao = alfabeto.find(letra_minuscula)

            nova_posicao = (posicao - deslocamento) % 26
            nova_letra = alfabeto[nova_posicao]

            if letra.isupper():
                resultado += nova_letra.upper()
            else:
                resultado += nova_letra
        else:
            resultado += letra

    return resultado


# Testando a descriptografia com a frase gerada no exercício anterior
texto_cifrado = "vhqdgrpxfholoxduh12345!"
casas = 3
texto_original = descriptografar_cesar(texto_cifrado, casas)

print(f"Cifrado:   {texto_cifrado}")
print(f"Decifrado: {texto_original}")
```

## 24. Validação de CPF
Implemente uma função para validar um CPF brasileiro.

```python
# Crio a função 'validar_cpf' e uso .replace() para tirar os pontos e o hífen.
# Verifico se o texto ficou com 11 dígitos e descarto CPFs com números todos iguais.
# Calculo o 1º dígito multiplicando os 9 primeiros números por um peso que começa em 10 e diminui.
# Calculo o 2º dígito multiplicando os 10 primeiros números por um peso que começa em 11 e diminui.
# Comparo se os dígitos calculados batem com os dígitos finais (índices 9 e 10) e retorno True ou False.

def validar_cpf(cpf):

    cpf = cpf.replace('.', '').replace('-', '')


    if len(cpf) != 11 or cpf == cpf[0] * 11:
        return False


    soma = 0
    peso = 10
    for i in range(9):
        soma += int(cpf[i]) * peso
        peso -= 1

    resto = (soma * 10) % 11
    if resto == 10:
        resto = 0
    primeiro_digito = resto


    soma = 0
    peso = 11
    for i in range(10):
        soma += int(cpf[i]) * peso
        peso -= 1

    resto = (soma * 10) % 11
    if resto == 10:
        resto = 0
    segundo_digito = resto


    if primeiro_digito == int(cpf[9]) and segundo_digito == int(cpf[10]):
        return True
    else:
        return False


# Testando
cpf_usuario = input("Digite um CPF (ex: 123.456.789-00): ")

if validar_cpf(cpf_usuario):
    print("CPF Válido!")
else:
    print("CPF Inválido!")
```

```python
# Defino a palavra secreta.
# letras_certas começa vazia e vai acumulando as letras que o jogador acerta.
# Enquanto tiver tentativas:
#   palavra_atual começa vazia a cada volta e é construída com as letras descobertas (ou underscores).
#   Se completou a palavra, venceu.
#   Senão, pede uma letra.
#   Se a letra está na palavra, adiciono a letras_certas.
#   Se não, diminuo uma tentativa.
# Se perdeu todas as tentativas, mostra a palavra.

palavra_secreta = 'carro'
letras_certas = ''
tentativas = 10

while tentativas > 0:
    palavra_atual = ''

    for letra in palavra_secreta:
        if letra in letras_certas:
            palavra_atual += letra
        else:
            palavra_atual += '_'

    print(f'Palavra: {palavra_atual}')
    print(f'Tentativas: {tentativas}')


    if '_' not in palavra_atual:
        print(f'Venceu! A palavra era: {palavra_secreta}')
        break

    letra = input('Adivinhe uma letra: ').lower()

    if letra in palavra_secreta:
        print('Acertou!')
        letras_certas += letra
    else:
        print('Errou!')
        tentativas -= 1

if tentativas == 0:
    print(f'Perdeu! A palavra era: {palavra_secreta}')
```

## 31. Ordenação por Bolha
Implemente o algoritmo de ordenação por bolha (Bubble Sort).

```python
# Leio uma lista de números do usuário.
# Uso dois laços for aninhados: o externo repete n vezes, o interno percorre a lista.
# A cada iteração do interno, comparo números adjacentes.
# Se o primeiro for maior que o segundo, troco de posição .
# Após n voltas, a lista fica ordenada em ordem crescente.

numeros = []

for i in range(5):
    num = int(input('Digite um número: '))
    numeros.append(num)

n = len(numeros)

for i in range(n):
    for j in range(n - 1):
        if numeros[j] > numeros[j + 1]:
            # aqui faz a troca de dois números de posição
            numeros[j], numeros[j + 1] = numeros[j + 1], numeros[j]

print(f'Lista ordenada: {numeros}')
```

## 32. Jogo da Velha
Crie um jogo da velha para dois jogadores.

```python
# Crio uma matriz 3x3 para o tabuleiro.
# Dois jogadores alternam entre 'X' e 'O', digitando linha e coluna.
# A cada rodada, mostro o tabuleiro usando join() para montar as linhas visualmente: ['X', ' ', 'O'] vira 'X | | O'.
# Verifico se ganhou (3 iguais em linha, coluna ou diagonal).
# Se o tabuleiro encher sem vencedor, é empate.

tabuleiro = [[' ', ' ', ' '],
             [' ', ' ', ' '],
             [' ', ' ', ' ']]

jogador = 'X'

while True:
    # mostra tabuleiro
    for linha in tabuleiro:
        print(' | '.join(linha))
        print('-' * 9)

    # pede jogada
    pos = int(input(f'Jogador {jogador} - digite posição (0-8): '))
    linha = pos // 3
    coluna = pos % 3

    if tabuleiro[linha][coluna] != ' ':
        print('Posição ocupada!')
        continue

    tabuleiro[linha][coluna] = jogador

    # verifica linhas
    ganhou = False
    for l in tabuleiro:
        if l[0] == l[1] == l[2] != ' ':
            ganhou = True

    # verifica colunas
    if not ganhou:
        for col in range(3):
            if tabuleiro[0][col] == tabuleiro[1][col] == tabuleiro[2][col] != ' ':
                ganhou = True

    # verifica diagonais
    if not ganhou:
        if tabuleiro[0][0] == tabuleiro[1][1] == tabuleiro[2][2] != ' ':
            ganhou = True
    if not ganhou:
        if tabuleiro[0][2] == tabuleiro[1][1] == tabuleiro[2][0] != ' ':
            ganhou = True

    if ganhou:
        for linha in tabuleiro:
            print(' | '.join(linha))
        print(f'Jogador {jogador} venceu!')
        break

    # verifica empate
    cheio = True
    for linha in tabuleiro:
        if ' ' in linha:
            cheio = False

    if cheio:
        for linha in tabuleiro:
            print(' | '.join(linha))
        print('Empate!')
        break

    # alterna jogador
    jogador = 'O' if jogador == 'X' else 'X'
```

## 33. Verificação de Anagramas
Verifique se duas palavras fornecidas são anagramas uma da outra.

```python
# Leio duas palavras e converto para minúsculas.
# Crio dois dicionários para contar a frequência de cada letra.
# Percorro cada palavra. Se a letra já está no dicionário, somo 1, senão crio com 1.
# Se os dois dicionários forem iguais, são anagramas.

palavra1 = input('Digite a primeira palavra: ').lower()
palavra2 = input('Digite a segunda palavra: ').lower()

contagem1 = {}
contagem2 = {}

for letra in palavra1:
    if letra in contagem1:
        contagem1[letra] = contagem1[letra] + 1
    else:
        contagem1[letra] = 1

for letra in palavra2:
    if letra in contagem2:
        contagem2[letra] = contagem2[letra] + 1
    else:
        contagem2[letra] = 1

if contagem1 == contagem2:
    print('São anagramas!')
else:
    print('Não são anagramas.')
```

## 37. Funções Recursivas
Implemente uma função recursiva para calcular o fatorial de um número.

```python
# Defino uma função que chama a si mesma (recursiva).
# Caso base: se n é 0 ou 1, retorna 1 (para a recursão).
# Caso recursivo: retorna n multiplicado pelo fatorial de n-1.
# A função vai descendo: fatorial(5) → fatorial(4) → fatorial(3) → até finalizar no 1

def fatorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * fatorial(n - 1)

numero = int(input('Digite um número: '))
print(f'Fatorial de {numero} é {fatorial(numero)}')
```

## 40. Lista de Compras
Crie um programa para gerenciar uma lista de compras: adicionar, remover e listar itens.

```python
# Crio uma lista vazia para guardar os itens.
# Uso um menu com while True que repete até o usuário escolher sair.
# Opções: 1=adicionar item, 2=remover item, 3=listar todos, 4=sair.
# append() adiciona itens, remove() tira itens da lista.
# Para exibir a lista numerada, pesquisei e encontrei enumerate(), que é uma função nativa do Python.
# enumerate() percorre a lista e devolve tanto o índice quanto o valor de cada elemento.
# Isso permite numerar a exibição sem precisar calcular manualmente.

lista_compras = []

while True:
    print('\n--- Lista de Compras ---')
    print('1. Adicionar item')
    print('2. Remover item')
    print('3. Listar itens')
    print('4. Sair')

    opcao = input('Escolha uma opção: ')

    if opcao == '1':
        item = input('Digite o item: ')
        lista_compras.append(item)
        print(f'"{item}" adicionado.')

    elif opcao == '2':
        item = input('Qual item remover? ')
        if item in lista_compras:
            lista_compras.remove(item)
            print(f'"{item}" removido.')
        else:
            print('Item não encontrado.')

    elif opcao == '3':
        if lista_compras:
            print('Itens da lista:')
            for i, item in enumerate(lista_compras, 1):
                print(f'{i}. {item}')
        else:
            print('Lista vazia.')

    elif opcao == '4':
        print('Até logo!')
        break

    else:
        print('Opção inválida.')
```

## 42. Números Primos até N
Gere todos os números primos até um número n fornecido.

```python
# Leio um número n como limite.
# Crio uma função para verificar se um número é primo.
# Percorro de 2 até n e testo cada número com a função.
# Adiciono os primos a uma lista.

def eh_primo(num):
    if num < 2:
        return False
    for i in range(2, num):
        if num % i == 0:
            return False
    return True

n = int(input('Digite o limite: '))
primos = []

for num in range(2, n + 1):
    if eh_primo(num):
        primos.append(num)

print(f'Números primos até {n}: {primos}')
```

## 46. Calcular Média e Desvio Padrão
Calcule a média e o desvio padrão de uma lista de números.

```python
# Leio números do usuário e adiciono a uma lista.
# Calculo a média dividindo a soma pela quantidade.
# Para calcular o desvio padrão: calculo a diferença de cada número com a média.
# Elevo cada diferença ao quadrado e somo todas.
# Divido a soma de quadrados pela quantidade e tiro a raiz quadrada com math.sqrt().

import math

numeros = []

for i in range(5):
    num = float(input('Digite um número: '))
    numeros.append(num)

media = sum(numeros) / len(numeros)

soma_quadrados = 0
for n in numeros:
    diferenca = n - media
    soma_quadrados += diferenca ** 2

desvio_padrao = math.sqrt(soma_quadrados / len(numeros))

print(f'Média: {media:.2f}')
print(f'Desvio padrão: {desvio_padrao:.2f}')
```

## 48. Ordenação por Inserção
Implemente o algoritmo de ordenação por inserção (Insertion Sort).

```python
# Leio números do usuário e adiciono a uma lista.
# Percorro a partir do índice 1 (segundo elemento).
# Para cada número (chave), comparo com os anteriores (j = i - 1).
# Enquanto houver números maiores que chave à esquerda, desloco eles para a direita.
# Quando encontro a posição certa ou j fica negativo, insiro chave naquela posição.
# Ao fim, a lista fica ordenada.

numeros = []

for i in range(5):
    num = int(input('Digite um número: '))
    numeros.append(num)

for i in range(1, len(numeros)):
    chave = numeros[i]
    j = i - 1

    while j >= 0 and numeros[j] > chave:
        numeros[j + 1] = numeros[j]
        j -= 1

    numeros[j + 1] = chave

print(f'Lista ordenada: {numeros}')
```

## 49. Soma dos Dígitos
Calcule a soma dos dígitos de um número inteiro fornecido.

```python
# Leio um número inteiro do usuário.
# Inicializo uma variável soma em 0.
# Enquanto o número for maior que 0:
#   Isolo: pego o último dígito com % 10.
#   Somo: adiciono esse dígito à soma.
#   Reduzo: divido por 10 com // 10 para remover o último dígito.
#   Repito: o while continua até o número chegar a 0.

numero = int(input('Digite um número: '))
soma = 0

while numero > 0:
    soma += numero % 10
    numero //= 10

print(f'Soma dos dígitos: {soma}')
```
